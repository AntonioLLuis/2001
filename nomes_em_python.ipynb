```python
# Exibe um título simples no console para organizar a saída
print("=== CADASTRO DE NOMES ===")

# Cria uma lista vazia chamada 'nomes' para armazenar os textos que o usuário digitar
nomes = []

# Inicia um laço que vai repetir exatamente 5 vezes (o range(5) gera os números 0, 1, 2, 3 e 4)
for i in range(5):
    # Solicita a entrada do usuário. O f-string "f{i + 1}" ajusta a contagem para o usuário ver de 1 a 5
    nome_digitado = input(f"Digite o  {i + 1}º nome: ")

    # Pega o nome digitado e o adiciona ao final da lista 'nomes' usando o método .append()
    nomes.append(nome_digitado)

# Imprime uma linha em branco (caractere de escape \n) para saltar uma linha no terminal
print("\n")

# Exibe um cabeçalho para a seção de saída de dados
print("=== MOSTRANDO A LISTA COMPLETA ===")

# Inicia outro laço para percorrer a lista e mostrar os nomes guardados
for i in range(5):
    # Acessa o valor na posição 'i' da lista nomes[i] e exibe na tela
    print(f"No  {i + 1}º nome digitado temos: {nomes[i]} ")
```
